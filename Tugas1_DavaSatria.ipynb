## 4. Menghapus data duplikat

**Input:** dataset mentah yang memiliki baris duplikat.  
**Proses:** mendeteksi dan menghapus baris yang seluruh nilainya sama dengan `drop_duplicates()`.  
**Output:** dataset tanpa duplikasi baris identik.

Catatan: penghapusan hanya dilakukan pada baris yang identik seluruh kolomnya. Duplikasi ID yang tidak identik perlu pemeriksaan bisnis terpisah.

```python
data = data.drop_duplicates().copy()
print('Ukuran setelah menghapus duplikat:', data.shape)
print('Sisa duplikat:', data.duplicated().sum())
```

## 5. Menangani inkonsistensi kategori dan typo

**Input:** kolom kategorik dengan variasi kapitalisasi, spasi, singkatan, dan salah eja.  
**Proses:** melakukan trimming spasi dan pemetaan nilai yang diketahui ke label baku.  
**Output:** kategori lebih konsisten.

Pemetaan dilakukan berdasarkan konteks nilai yang memang jelas, misalnya `L` menjadi `Laki-laki`, `P` menjadi `Perempuan`, dan `Sistem Informsi` menjadi `Sistem Informasi`.

```python
data['Program_Studi'] = data['Program_Studi'].astype('string').str.lower()
data['Program_Studi'] = data['Program_Studi'].replace({
    'sistem informasi': 'Sistem Informasi',
    'Sistem Informsi': 'Sistem Informasi',
    'Informatika': 'Informatika'
})

data['Jenis_Kelamin'] = data['Jenis_Kelamin'].astype('string').str.strip()
data['Jenis_Kelamin'] = data['Jenis_Kelamin'].replace({
    'L': 'Laki-laki', 'l': 'Laki-laki',
    'P': 'Perempuan', 'p': 'Perempuan'
})

data['Status_Mahasiswa'] = data['Status_Mahasiswa'].astype('string').str.strip()
data['Status_Mahasiswa'] = data['Status_Mahasiswa'].replace({
    'aktif': 'Aktif',
    'Nonaktif': 'Tidak Aktif',
    'nonaktif': 'Tidak Aktif'
})

data['Jenis_Kelas'] = data['Jenis_Kelas'].astype('string').str.strip()
data['Jenis_Kelas'] = data['Jenis_Kelas'].replace({
    'reguler': 'Reguler',
    'REGULER': 'Reguler',
    'Karyawan ': 'Karyawan',
    'reguler/karyawan': 'Reguler/Karyawan'
})

for col in ['Program_Studi', 'Jenis_Kelamin', 'Status_Mahasiswa', 'Jenis_Kelas']:
    print(col, data[col].value_counts(dropna=False).to_dict())
```

## 6. Validasi nilai numerik dan identifikasi nilai tidak valid

**Input:** variabel numerik yang mungkin memiliki nilai di luar domain.  
**Proses:** konversi tipe numerik dan pemeriksaan batas logis yang ditetapkan untuk simulasi ini:
- Usia: 17–60 tahun.
- Semester: 1–14.
- Jumlah SKS: 0–26.
- IPK: 0–4.
- Persentase kehadiran: 0–100.
- UKT: Rp0–Rp20.000.000.

Nilai di luar batas dianggap tidak valid untuk konteks dataset ini dan diubah menjadi missing value agar dapat ditangani secara transparan. Batas ini adalah asumsi operasional untuk latihan, sehingga perlu disesuaikan jika institusi memiliki aturan berbeda.

**Output:** nilai tidak valid ditandai sebagai missing value.

```python
numeric_cols = ['Usia', 'Semester', 'Jumlah_SKS', 'IPK',
                'Persentase_Kehadiran', 'UKT_Rp']
for col in numeric_cols:
    data[col] = pd.to_numeric(data[col], errors='coerce')

valid_ranges = {
    'Usia': (17, 60),
    'Semester': (1, 14),
    'Jumlah_SKS': (0, 26),
    'IPK': (0, 4),
    'Persentase_Kehadiran': (0, 100),
    'UKT_Rp': (0, 20_000_000)
}

invalid_counts = {}
for col, (lower, upper) in valid_ranges.items():
    invalid_mask = data[col].notna() & ~data[col].between(lower, upper)
    invalid_counts[col] = int(invalid_mask.sum())
    data.loc[invalid_mask, col] = np.nan

print('Jumlah nilai tidak valid per variabel:')
display(pd.Series(invalid_counts, name='Jumlah nilai tidak valid'))
```

## 7. Identifikasi outlier dengan metode IQR

**Input:** nilai numerik pada dataset sebelum imputasi.  
**Proses:** menghitung Q1, Q3, IQR, serta batas bawah dan atas (`Q1 - 1.5×IQR`, `Q3 + 1.5×IQR`).  
**Output:** ringkasan kandidat outlier untuk ditinjau.

IQR digunakan sebagai alat deteksi, bukan alasan otomatis untuk menghapus data. Nilai ekstrem diperiksa kembali dengan batas logis/domain. Nilai yang melanggar batas domain ditangani pada tahap validasi sebelumnya.

```python
iqr_rows = []
for col in numeric_cols:
    series = data[col].dropna()
    if len(series) == 0:
        continue
    q1, q3 = series.quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    count = int(((series < lower) | (series > upper)).sum())
    iqr_rows.append([col, q1, q3, iqr, lower, upper, count])

iqr_summary = pd.DataFrame(iqr_rows, columns=[
    'Variabel', 'Q1', 'Q3', 'IQR', 'Batas Bawah', 'Batas Atas', 'Kandidat Outlier'
])
display(iqr_summary)
```

## 8. Menangani format tanggal

**Input:** kolom `Tanggal_Daftar` dengan beberapa format tanggal.  
**Proses:** mencoba beberapa format yang diketahui. Tanggal yang tidak valid (misalnya 31/02/2024) menjadi `NaT`. Baris dengan tanggal pendaftaran tidak valid dihapus karena tanggal sebenarnya tidak dapat disimpulkan secara bertanggung jawab.  
**Output:** tanggal valid dalam format ISO `YYYY-MM-DD`.

```python
from datetime import datetime

date_formats = ['%Y-%m-%d', '%d/%m/%Y', '%d-%m-%Y', '%Y/%m/%d']

def parse_date(value):
    if pd.isna(value):
        return pd.NaT
    value = str(value).strip()
    for fmt in date_formats:
        try:
            return pd.Timestamp(datetime.strptime(value, fmt))
        except ValueError:
            continue
    return pd.NaT

data['Tanggal_Daftar'] = data['Tanggal_Daftar'].apply(parse_date)
invalid_date_count = int(data['Tanggal_Daftar'].isna().sum())
print('Tanggal kosong/tidak valid:', invalid_date_count)
data = data.dropna(subset=['Tanggal_Daftar']).copy()
data['Tanggal_Daftar'] = data['Tanggal_Daftar'].dt.strftime('%Y-%m-%d')
print('Ukuran setelah validasi tanggal:', data.shape)
```

## 9. Menangani missing value dengan statistik deskriptif

**Input:** missing value asli dan missing value hasil validasi nilai tidak valid.  
**Proses:**
- Variabel numerik: imputasi median karena lebih tahan terhadap nilai ekstrem dibanding mean.
- Variabel kategorik: imputasi modus karena mengisi nilai dengan kategori paling sering muncul.

Nilai median/modus dihitung dari data yang tersedia setelah duplikasi, inkonsistensi, dan nilai tidak valid ditangani.

**Output:** nilai kosong pada variabel yang ditangani terisi.

```python
numeric_imputation = {}
for col in ['Usia', 'Jumlah_SKS', 'IPK', 'Persentase_Kehadiran', 'UKT_Rp']:
    median_value = data[col].median()
    numeric_imputation[col] = median_value
    data[col] = data[col].fillna(median_value)

categorical_imputation = {}
for col in ['Program_Studi', 'Jenis_Kelamin', 'Status_Mahasiswa', 'Jenis_Kelas']:
    mode_value = data[col].mode(dropna=True).iloc[0]
    categorical_imputation[col] = mode_value
    data[col] = data[col].fillna(mode_value)

print('Median numerik:', numeric_imputation)
print('Modus kategorik:', categorical_imputation)
print('Missing value setelah imputasi:')
display(data.isna().sum())
```

## 10. Standardisasi tipe data dan format akhir

**Input:** data yang telah divalidasi dan diimputasi.  
**Proses:** mengembalikan variabel diskrit ke tipe integer, membulatkan IPK dan kehadiran, serta menyeragamkan tanggal.  
**Output:** tipe data lebih konsisten untuk analisis.

```python
for col in ['Usia', 'Semester', 'Jumlah_SKS']:
    data[col] = data[col].round().astype(int)
data['IPK'] = data['IPK'].round(2)
data['Persentase_Kehadiran'] = data['Persentase_Kehadiran'].round(1)
data['UKT_Rp'] = data['UKT_Rp'].round().astype('int64')

display(data.dtypes)
display(data.head())
```
